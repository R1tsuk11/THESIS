This notebook provides a comprehensive overview of the class structure and relationships in the Arami Waray language learning application. The diagram shows the main classes, their attributes, methods, and relationships.

## Overview of Main Components

The application consists of several key components:

1. **User Interface (Flet)**: Manages the UI using the Flet framework
2. **Learning Engine Components**:
   - **BKT Engine**: Bayesian Knowledge Tracing for student knowledge modeling
   - **LSTM Engine**: Long Short-Term Memory neural networks for proficiency prediction
   - **SuperMemo Engine**: Spaced repetition algorithm for vocabulary scheduling
3. **Speech Recognition**: For pronunciation assessment
4. **Data Models**: User, Module, Level, Question, etc.

## Class Relationships

This section shows how the various classes in the Arami application relate to each other. The diagram illustrates:

1. **Composition Relationships** (has-a): Shown with solid lines and diamond arrowheads. These represent strong ownership relationships where one class contains instances of another class.

2. **Usage Relationships** (uses): Shown with dashed lines. These represent dependencies where one class uses the functionality of another class.

The complete class relationship diagram provides an overview of how all components interact within the system architecture.

```python
# Class relationships description
try:
    # Try to create diagram if Graphviz executables are available
    relationship_diagram = graphviz.Digraph('Class_Relationships', comment='Class Relationships')
    
    # Set graph attributes correctly
    relationship_diagram.graph_attr.update({
        'rankdir': 'TB',        # Top to bottom layout
        'splines': 'ortho',     # Orthogonal lines
        'nodesep': '0.8',       # More space between nodes
        'ranksep': '1.0',       # More space between ranks
        'fontsize': '12'        # Larger font
    })
    
    # Set default node attributes
    relationship_diagram.node_attr.update({
        'shape': 'box',         # Use box shape for better readability
        'style': 'filled',
        'fontsize': '12'
    })

    # Define nodes for all classes with better formatting
    relationship_diagram.node('User', 'User', fillcolor='lightblue')
    relationship_diagram.node('Module', 'Module', fillcolor='lightgreen')
    relationship_diagram.node('Level', 'Level', fillcolor='lightgreen')
    relationship_diagram.node('Question', 'Question', fillcolor='lightgreen')
    relationship_diagram.node('ChapterTest', 'ChapterTest', fillcolor='lightgreen')
    relationship_diagram.node('Achievements', 'Achievements', fillcolor='lightyellow')
    relationship_diagram.node('BKT_Engine', 'CustomBKTPredictor', fillcolor='lightpink')
    relationship_diagram.node('LSTM_Engine', 'LSTM Engine', fillcolor='lightpink')
    relationship_diagram.node('SuperMemo_Engine', 'SuperMemo Engine', fillcolor='lightpink')

    # Define the relationships with labels
    # Composition relationships (strong ownership)
    relationship_diagram.edge('User', 'Module', label='has many', arrowhead='diamond', color='blue', fontcolor='blue')
    relationship_diagram.edge('Module', 'Level', label='has many', arrowhead='diamond', color='blue', fontcolor='blue')
    relationship_diagram.edge('Module', 'ChapterTest', label='has one', arrowhead='diamond', color='blue', fontcolor='blue')
    relationship_diagram.edge('Level', 'Question', label='has many', arrowhead='diamond', color='blue', fontcolor='blue')
    relationship_diagram.edge('User', 'Achievements', label='has many', arrowhead='diamond', color='blue', fontcolor='blue')

    # Usage relationships
    relationship_diagram.edge('User', 'BKT_Engine', label='uses', style='dashed', color='red', fontcolor='red')
    relationship_diagram.edge('User', 'LSTM_Engine', label='uses', style='dashed', color='red', fontcolor='red')
    relationship_diagram.edge('User', 'SuperMemo_Engine', label='uses', style='dashed', color='red', fontcolor='red')
    relationship_diagram.edge('Module', 'SuperMemo_Engine', label='schedules review', style='dashed', color='red', fontcolor='red')
    relationship_diagram.edge('Question', 'BKT_Engine', label='tracked by', style='dashed', color='red', fontcolor='red')

    # Render and save the diagram with higher DPI for better quality
    relationship_diagram.render('class_relationships', format='png', cleanup=True)
    # Display the diagram
    display(Image('class_relationships.png'))
    print("Class relationship diagram created successfully!")
except Exception as e:
    print(f"Could not create graphical diagram: {e}")
    # Use text description as fallback
    display(Markdown("### Class Relationships"))
    
    display(Markdown("#### Composition Relationships (has-a)"))
    composition = [
        "User **has many** Module",
        "Module **has many** Level",
        "Module **has one** ChapterTest",
        "Level **has many** Question",
        "User **has many** Achievements"
    ]
    for rel in composition:
        display(Markdown(f"- {rel}"))
    
    display(Markdown("#### Usage Relationships (uses)"))
    usage = [
        "User **uses** CustomBKTPredictor",
        "User **uses** LSTM Engine",
        "User **uses** SuperMemo Engine",
        "Module **schedules review** with SuperMemo Engine",
        "Question **tracked by** CustomBKTPredictor"
    ]
    for rel in usage:
        display(Markdown(f"- {rel}"))
    
    display(Markdown("---"))
```

## Summary

The Arami Waray language learning application is structured around a core set of classes that handle different aspects of the learning process:

1. **Data Model Classes**: 
   - `User`: Central entity that manages the user's state, progress, and relationships to other components
   - `Module`: Represents learning modules containing levels and chapter tests
   - `Level`: Contains questions and tracks completion status
   - `Question`: Represents individual learning items with various properties
   - `Achievements`: Tracks user accomplishments
   - `ChapterTest`: Assessment for entire modules

2. **Learning Engine Components**:
   - `CustomBKTPredictor`: Implements Bayesian Knowledge Tracing to model knowledge state
   - `LSTM Engine`: Uses neural networks to predict overall proficiency
   - `SuperMemo Engine`: Implements spaced repetition algorithm for optimizing review schedules

3. **Relationships**:
   - The User class has the central position, owning modules, achievements, and library items
   - Modules contain multiple levels and a chapter test
   - Levels contain multiple questions
   - The learning engines interact with user data to optimize the learning experience

This architecture enables adaptive learning, personalized scheduling, and accurate progress tracking throughout the application.
